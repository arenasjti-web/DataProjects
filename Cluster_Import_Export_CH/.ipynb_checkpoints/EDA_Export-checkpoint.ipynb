# Distribución de variables categóricas

```python
list( df_export.select_dtypes(include="str").columns)
```

```python
categorical_cols = ['rubro_empresa', 'empresa', 'pais', 'via_transporte']
```

```python
distribuciones = []

for col in categorical_cols:
    conteo = df_export[col].value_counts(dropna=False)

    porcentaje = df_export[col].value_counts(
        normalize=True,
        dropna=False
    ) * 100

    temp = pd.DataFrame({
        "variable": col,
        "categoria": conteo.index,
        "cantidad": conteo.values,
        "porcentaje": porcentaje.values
    })

    distribuciones.append(temp)

distribucion_categoricas = pd.concat(
    distribuciones,
    ignore_index=True
)
```

```python
categorical_cols = ['rubro_empresa', 'empresa', 'pais', 'via_transporte']
distribuciones = []
for col in categorical_cols:
    datos = (
        distribucion_categoricas[
            distribucion_categoricas["variable"] == col
        ]
        .sort_values("porcentaje", ascending=True)
    )

    plt.figure(figsize=(10, max(5, len(datos) * 0.3)))

    sns.barplot(
        data=datos,
        x="porcentaje",
        y="categoria"
    )

    plt.xlabel("Porcentaje (%)")
    plt.ylabel(col)
    plt.title(f"Distribución de {col}")
    plt.tight_layout()
    plt.show()
```

# Outliers

```python
n_cols = 3
n_rows = (len(numeric_cols) + n_cols - 1) // n_cols

outliers = []
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 3 * n_rows))
axes = axes.flatten()

for ax, col in zip(axes, numeric_cols):

    Q1 = df_export[col].quantile(0.25)
    Q3 = df_export[col].quantile(0.75)
    IQR = Q3 - Q1
    
    outliers.append({
        "col":col,
        "data": df_export[(df_export[col] < Q1 - 1.5 * IQR) | 
            (df_export[col] > Q3 + 1.5 * IQR)]
    })
    
    data = np.log1p(df_export[col])
    sns.boxplot(data, ax=ax)
    ax.set_title(col)

for ax in axes[len(numeric_cols):]:
    fig.delaxes(ax)

plt.tight_layout()
plt.show()
```

```python
Q1 = df_export["total_operacion"].quantile(0.25)
Q3 = df_export["total_operacion"].quantile(0.75)
IQR = Q3 - Q1

df_export["es_outlier_total_operacion"] = (
    df_export["total_operacion"] < Q1 - 1.5 * IQR
) | (
    df_export["total_operacion"] > Q3 + 1.5 * IQR
)
```

```python
df_export.groupby("es_outlier_total_operacion")[
    ["total_operacion", "peso_total", "cantidad_operacion"]
].median()
```

```python
df_export["es_outlier_total_operacion"].value_counts(normalize=True)
```
