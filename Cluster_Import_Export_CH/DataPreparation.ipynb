```python
#Con las fuentes separadas este campo ya no hace falta
df_export = df_export.drop(columns=["tipo_operacion"])
df_import = df_import.drop(columns=["tipo_operacion"])
```

```python
#Mismas columnas para ambas fuentes

#No se incluyen columnas que no representan medidas ( ej: capitulo_arancel )
numerical_cols = [
    "total_operacion",
    "cantidad_operacion",
    "ticket_promedio",
    "peso_total",
]

categorical_cols = [
    "rubro_empresa",
    "empresa",
    "descripcion_arancel",#Quitar antes de entrenar modelo
    "pais",
    "via_transporte"
]
```

#### Observación: Nombres de columnas ya se encuentran normalizadas. No hay espacios en blancos ni carácteres especiales

# Normalizar valores Categóricos

```python
def formatCategorical(df,categorical_cols):
    for c in categorical_cols:
        df[c] = df[c].str.lower().str.replace(' ', '_')
```

```python
formatCategorical(df_import,categorical_cols)
formatCategorical(df_export,categorical_cols)
```
